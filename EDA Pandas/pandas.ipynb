### `Group by and Aggregating`

```python
import pandas as pd
```

```python
df = pd.read_csv(r"C:\Users\ahmad\Downloads\Flavors.csv")
df
```

```python
#group base flavor
group_by_frame = df.groupby('Base Flavor')
```

```python
#average 
group_by_frame.mean(numeric_only=True)
```

```python
# count
df.groupby('Base Flavor').count()
```

```python
# min & max
df.groupby('Base Flavor').min()
```

```python
df.groupby('Base Flavor').max()
```

```python
# sum
df.groupby('Base Flavor').sum(numeric_only = True)
```

```python
# aggregate flavor rating
df.groupby('Base Flavor').agg({'Flavor Rating':['mean','max','count','sum']})
```

```python
# aggregate flavor rating and texture rating
df.groupby('Base Flavor').agg({'Flavor Rating':['mean','max','count','sum'], 'Texture Rating':['mean','max','count','sum']})
```

```python
df
```

```python
# group base flavor & liked
df.groupby(['Base Flavor','Liked']).mean(numeric_only = True)
```

```python
df.groupby(['Base Flavor','Liked']).agg({'Flavor Rating':['mean','max','count','sum']})
```

```python
# describe
df.groupby('Base Flavor').describe()
```

### `Merging Data Frame`

```python
import pandas as pd
```

```python
df1 = pd.read_csv(r"C:\Users\ahmad\Downloads\LOTR.csv")
df1
```

```python
df2 = pd.read_csv(r"C:\Users\ahmad\Downloads\LOTR 2.csv")
df2
```

```python
# merge
# inner join = default
df1.merge(df2)
```

```python
# inner join 
df1.merge(df2, how = 'inner', on = 'FellowshipID')
```

```python
# inner join 
df1.merge(df2, how = 'inner', on = ['FellowshipID','FirstName'])
```

```python
# outer join
df1.merge(df2, how = 'outer')
```

```python
# left join
df1.merge(df2, how = 'left')
```

```python
# right join
df1.merge(df2, how = 'right')
```

```python
# cross join
df1.merge(df2, how = 'cross')
```

```python
# join function
df1.join(df2, on = 'FellowshipID', how = 'outer', lsuffix = '_Left',rsuffix = '_Right')
```

```python
# index
df4 = df1.set_index('FellowshipID').join(df2.set_index('FellowshipID'), lsuffix = '_Left',rsuffix = '_Right', how = 'outer')
df4
```

```python
# concatenate
pd.concat([df1,df2])
```

```python
pd.concat([df1,df2], join = 'outer', axis = 1)
```

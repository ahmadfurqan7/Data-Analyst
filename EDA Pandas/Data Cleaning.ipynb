# *DATA CLEANING*

```python
# import library
import pandas as pd
```

```python
# load dataset
df = pd.read_excel(r"C:\Users\ahmad\Downloads\Customer Call List.xlsx")
df
```

```python
# Drop the duplicate
df = df.drop_duplicates()
df
```

```python
# Remove not useful column 
df = df.drop(columns = "Not_Useful_Column")
df
```

```python
# Fix the values of Column
# last_Name column
df["Last_Name"].str.strip() #check error on the values
```

```python
# df["Last_Name"] = df["Last_Name"].str.lstrip("...") # remove dot
# df["Last_Name"] = df["Last_Name"].str.lstrip("/") # remove slash
# df["Last_Name"] = df["Last_Name"].str.lstrip("_") # remove underscore
# df["Last_Name"] = df["Last_Name"].str.rstrip("_") # remove underscore
# This is for one by one
```

```python
# for multiple change with single line
df["Last_Name"].str.strip("123._/") # combine all error values
```

```python
# input to variables and check the values
df["Last_Name"] = df["Last_Name"].str.strip("123._/")
df
```

```python
# fix the Phone_Number column values
df["Phone_Number"].str.replace('[^a-zA-Z0-9]','') #change the non numeric values
```

```python
# remove non numeric values
df["Phone_Number"] = df["Phone_Number"].str.replace('[^a-zA-Z0-9]','',regex = True)
df
```

```python
# change type of values to string
df["Phone_Number"].apply(lambda x : str(x))
```

```python
# input to variables
df["Phone_Number"] = df["Phone_Number"].apply(lambda x : str(x))
```

```python
# using lambda to change format phone number values
df["Phone_Number"].apply(lambda x : x[0:3] + '-' + x[3:6] + '-' + x[6:10]) # pick one format calues
```

```python
# input to variables
df["Phone_Number"] = df["Phone_Number"].apply(lambda x : x[0:3] + '-' + x[3:6] + '-' + x[6:10])
df
```

```python
# get rid off nan-- and NA--
df["Phone_Number"] = df["Phone_Number"].str.replace('nan--','')
df["Phone_Number"] = df["Phone_Number"].str.replace('Na--','')
```

```python
df
```

```python
# split the address column
df["Address"].str.split(',',n=2, expand = True) #split with 2 commas
```

```python
# Create column to data
df[["Street_Address","State","Zip_Code"]] = df["Address"].str.split(',',n=2, expand = True)
df
```

```python
# replace values of Paying Customer and Do not contact column
df["Paying Customer"] = df["Paying Customer"].str.replace("Yes","Y")  #Yes to Y
df["Paying Customer"] = df["Paying Customer"].str.replace("No","N")  #No to N
df["Do_Not_Contact"] = df["Do_Not_Contact"].str.replace("Yes","Y")  #Yes to Y
df["Do_Not_Contact"] = df["Do_Not_Contact"].str.replace("No","N")  #No to N
df
```

```python
# Remove N/A values and input to dataset
df = df.replace("N/a",'')
```

```python
# The values NaN is empty so fill it
df = df.fillna('')
df
```

```python
# Filter
# list the values with the contact and phone number
for x in df.index: # using loop function
    if df.loc[x,"Do_Not_Contact"] == "Y":
        df.drop(x, inplace = True)
df
```

```python
for x in df.index: # using loop function
    if df.loc[x,"Phone_Number"] == "": # remove row with blank values
        df.drop(x, inplace = True)
df

# another way to drop null values
# df = df.dropna(subset="Phone_Number"),inplace = True
```

```python
# create new index
df = df.reset_index(drop=True)
df
```
